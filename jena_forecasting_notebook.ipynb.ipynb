```python

from pathlib import Path

path = Path("data_utils.py")
text = path.read_text()

old_code = """def load_dataframe(csv_path: str, subsample_every: int = 6) -> pd.DataFrame:
    \"\"\"Loads the CSV. The data is every 10 minutes; subsample_every=6 means hourly sampling
    (the standard method in TF tutorials to reduce sequence length and noise).
    \"\"\"
    df = pd.read_csv(csv_path)"""

# New code with the ability to detect the file inside a folder
new_code = """def load_dataframe(csv_path: str, subsample_every: int = 6) -> pd.DataFrame:
    \"\"\"Loads the CSV. The data is every 10 minutes; subsample_every=6 means hourly sampling
    (the standard method in TF tutorials to reduce sequence length and noise).
    \"\"\"
    import os, glob
    if os.path.isdir(csv_path):
        candidates = glob.glob(os.path.join(csv_path, "**/*.csv"), recursive=True)
        if not candidates: raise FileNotFoundError(f"No CSV in {csv_path}")
        csv_path = candidates[0]
    print(f"--- Loading data from: {csv_path} ---")
    df = pd.read_csv(csv_path)"""

if old_code in text:
    text = text.replace(old_code, new_code)
    path.write_text(text)
    print("data_utils.py was modified successfully!")
else:
    lines = text.splitlines()
    if "df = pd.read_csv(csv_path)" in lines[29]: # line 30 (index 29)
        lines[29] = """    import os, glob
    if os.path.isdir(csv_path):
        csv_path = glob.glob(os.path.join(csv_path, "**/*.csv"), recursive=True)[0]
    print(f"--- Loading data from: {csv_path} ---")
    df = pd.read_csv(csv_path)"""
        path.write_text("\\n".join(lines))
        print("data_utils.py was modified via line number!")
    else:
        print("The requested code was not found. Please check the file content again.")
```
